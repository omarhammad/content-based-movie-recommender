```python
import pandas as pd
import requests
import time

movies_df = pd.read_csv('themoviedb.org.csv')

# Your TMDb API Key
api_key = '7f0c99fee6bd9740878aaa8d2c93ca86'

```

### Define Function to Get Movie Overview
Here, we define a function called `fetch_overview()` that takes a movie title and searches for it on The Movie Database (TMDb) using their API.


```python
def fetch_overview(title):
    url = f'https://api.themoviedb.org/3/search/movie?api_key={api_key}&query={requests.utils.quote(title)}'
    response = requests.get(url)
    if response.status_code == 200:
        results = response.json().get('results')
        if results:
            return results[0].get('overview')  # Return the first result's overview
    return None
```

### Fetch Overviews and Save the Data
Once we have all the overviews, we add them to our original table (`movies_df`) and save everything to a new CSV file named `themoviedb_with_overviews.csv`.


```python
overviews = []
counter = 0
for title in movies_df['title']:
    overview = fetch_overview(title)
    print(f'{counter}| {title} || {overview}')
    overviews.append(overview)
    counter += 1
    time.sleep(0.25)

# Add the overviews to the DataFrame
movies_df['overview'] = overviews

# Save the DataFrame with overviews to a new CSV file
output_file = 'themoviedb_with_overviews.csv'
movies_df.to_csv(output_file, index=False)

print(f"New file with overviews saved as '{output_file}'")
```
